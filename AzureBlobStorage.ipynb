```python
#uses azuredemo conda environment - Python 3.12.12
from dotenv import load_dotenv
import os
import io   
import pandas as pd
import numpy as np
```

## Microsoft Azure Blob Storage Client Library for Python

Azure Blob Storage is ideal for storing large amounts of unstructured data such as text or binary data at low cost. See https://pypi.org/project/azure-storage-blob for the Python client library. Python 3.9 or later is required to use this package - see https://github.com/Azure/azure-sdk-for-python/blob/main/doc/python_version_support_policy.md for more details. It is however safer to already start using 3.10 or higher ...

```python
from azure.storage.blob import BlobServiceClient, BlobClient,ContainerClient,__version__
__version__
```

## Loading Azure storage connection string
os.getenv will retrieve it standard from the environment variable which you can on Windows:Control Panel -> System -> Advanced system settings -> Environment Variables
By using python-dotenv - you can keep your connection string in a .env file and load it into your environment variables. This is useful for local development and testing.

```python
load_dotenv()  # Load environment variables from .env file
connectionstring = os.getenv('AZURE_STORAGE_CONNECTION_STRING')

blobcontainer = "demo"
blob_service_client = BlobServiceClient.from_connection_string(connectionstring)
container = blob_service_client.get_container_client(blobcontainer)
#print(container.url)
```

```python
#Create a sample dataframe
np_sample = np.array([[1,2],[3,4]])
df = pd.DataFrame(data=np_sample,index=["row1","row2"],columns=['column1','column2'])
df.head()
```

```python
#Create a csv from a dataframe and upload to blob storage
upload_data = io.StringIO()
upload_data = df.to_csv()
container.upload_blob(data=upload_data, name='dfdemo.csv', overwrite=True)
```

```python
blob_stream = io.BytesIO(container.get_blob_client(blob='dfdemo.csv').download_blob().readall())
dffromblob = pd.read_csv(blob_stream)
dffromblob.head()
```

```python
def save_blob(file_name,file_content):
        download_file_path = os.path.join('c:/temp',file_name)

        with open(download_file_path, "wb") as file:
            file.write(file_content)

```

```python
my_blobs = container.list_blobs()

for blob in my_blobs:
        print (blob.name)
        #Download the blob from Azure Storage
        bytes = container.get_blob_client(blob).download_blob().readall()
        save_blob(blob.name,bytes)
```
